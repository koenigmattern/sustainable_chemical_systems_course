# Installation instructions

* The course content was written in Python 3.12 in the VSCode editor
* You can find the course contents under `../suschemsys` in the Jupyter Notebook-files (.ipynb)
* The notebooks can be run locally wither on your machine and on Google colab (requires login with your google account). Instructions below.

## Local machine
In this [video](https://www.youtube.com/watch?time_continue=192&v=ljFwYKL6-1U&embeds_referring_euri=https%3A%2F%2Fwww.google.com%2F&source_ve_path=MTM5MTE3LDI4NjY2) you can find everything you need for installing Python, the editor and the package manager Anaconda (`conda`). Our coding example requires several python modules that we need to install. We use `conda` to create a virtual environment for the course with the name `suschemsys`. 

On your local machine, open your terminal and create a new conda environment from the .yml-file

```python
conda env create --name suschemsys --file=environment.yml
```

Then we simply need to activate the environment:

```python
conda activate suschemsys
```

Great! Now everything is ready on your local machine! :+1:
